### Гистограмма распределения значений popularity

```python
plt.figure(figsize=(10, 6))
plt.hist(df['popularity'], bins=20, color='skyblue', edgecolor='black')
plt.title('Распределение значений popularity')
plt.xlabel('Popularity')
plt.ylabel('Частота')
plt.xticks(range(0, 101, 10))
plt.grid(axis='y', alpha=0.75)
plt.show()
```

#### 1. Основная часть значений находится в диапазоне от 10 до 40
#### 2. Значения выше 60 встречаются значительно реже остальных

## <center>Контрольные вопросы</center>

### 1. Что такое DataFrame?
DataFrame — это табличная структура данных с именованными столбцами и индексами строк, предназначенная для удобного хранения, преобразования и анализа структурированных данных в аналитических и научных вычислениях.
### 2. Что такое признак (Feature)?
Признак (feature) — отдельная измеримая характеристика примера, по которой модель делает вывод. В таблице датасета признаки — это столбцы: возраст, сумма покупки, длина текста, время суток. От выбора признаков качество модели зависит сильнее, чем от алгоритма.
### 3. Чем среднее арифметическое отличается от медианы?
Среднее арифметическое (mean) рассчитывается как сумма всех значений выборки, деленная на их количество. Медиана (median) — это значение, которое делит упорядоченный по возрастанию набор данных пополам: 50% значений в выборке меньше медианы, а остальные 50% — больше. На медиану не влияют экстремально высокие или низкие значения.
### 4. Что показывает гистограмма?
Гистограмма — это графическое представление распределения числовых данных. Она демонстрирует, как часто значения признака попадают в определенные интервалы.
### 5. Для чего нужна фильтрация данных?
Фильтрация данных необходима для отбора подмножества строк из исходного набора данных, которые удовлетворяют заданным логическим условиям.
